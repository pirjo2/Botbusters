```python
import pandas as pd
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Network Science/project/bot_detection_data.csv')
```

```python
df.head()
```

```python

from sklearn.model_selection import KFold
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
import numpy as np
import pandas as pd # Ensure pandas is imported

# Drop the object type columns that are not suitable for direct input to XGBoost
# and were not explicitly processed in this block
X = df.drop(['Bot Label', 'User ID', 'Username', 'Tweet', 'Location', 'Created At', 'Hashtags'], axis=1)
y = df['Bot Label']

kf = KFold(n_splits=5, shuffle=True, random_state=42)
accuracies = []

```

```python


from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
# Add other models to the list
models = [
    ('XGBoost', XGBClassifier(use_label_encoder=False, eval_metric='logloss')),
    ('Logistic Regression', LogisticRegression(max_iter=1000)),
    ('Random Forest', RandomForestClassifier()),
    ('SVM', SVC())
]
results = {}
for name, model in models:
    accuracies = []
    for train_index, test_index in kf.split(X):
        X_train, X_test = X.iloc[train_index], X.iloc[test_index]
        y_train, y_test = y.iloc[train_index], y.iloc[test_index]
        # Ensure the columns in X_train and X_test are of supported dtypes
        X_train = X_train.astype({'Retweet Count': int, 'Mention Count': int, 'Follower Count': int, 'Verified': int})
        X_test = X_test.astype({'Retweet Count': int, 'Mention Count': int, 'Follower Count': int, 'Verified': int})
        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        accuracy = accuracy_score(y_test, y_pred)
        accuracies.append(accuracy)
    results[name] = {
        "accuracies": accuracies,
        "mean_accuracy": np.mean(accuracies),
        "std_accuracy": np.std(accuracies)
    }
for name, result in results.items():
    print(f"{name} Results:")
    print(f"Cross-validation accuracies: {result['accuracies']}")
    print(f"Mean accuracy: {result['mean_accuracy']}")
    print(f"Standard deviation of accuracy: {result['std_accuracy']}")
    print("-" * 20)
```
